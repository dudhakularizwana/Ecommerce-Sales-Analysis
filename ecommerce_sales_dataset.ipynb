```python
def recommend_action(row):
    if row["Total_Sales"] > 10000 and row["Customer_Rating"] > 4:
        return "PROMOTE HEAVILY"
    elif row["Total_Sales"] < 5000 and row["Customer_Rating"] < 3:
        return "CONSIDER DISCONTINUE"
    else:
        return "KEEP AS IS"
```

```python
df["Business_Action"] = df.apply(recommend_action, axis=1)
df[["Order_ID","Product_Category","Total_Sales","Customer_Rating","Business_Action"]]
```
